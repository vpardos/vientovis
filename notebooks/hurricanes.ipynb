```python
import pandas as pd
from hurdat2 import load_hurdat2, add_derived_columns

#cargar hurdat
population_df = pd.read_csv(r"C:\Users\espeo\OneDrive\Escritorio\vientovis\vientovis\vientovis\data\population\co-est2024-alldata.csv", encoding='cp1252')
hurdat2_df, hurricanes = load_hurdat2(r"C:\Users\espeo\OneDrive\Escritorio\vientovis\vientovis\vientovis\notebooks\hurricanes\hurdat2-1851-2025-091226.txt")
hurdat2_df = add_derived_columns(hurdat2_df)

hurdat2_df[(hurdat2_df["storm_name"] == "KATRINA") & (hurdat2_df["date"].str.contains("2005"))]

```

```python
Impactos = pd.read_csv(r"C:\Users\espeo\OneDrive\Escritorio\vientovis\vientovis\vientovis\data\impacts\storm_events_daily.csv")
Dano = pd.read_csv (r"C:\Users\espeo\OneDrive\Escritorio\vientovis\vientovis\vientovis\data\damage\nfip_daily_damage.csv")
Cantidad = pd.read_csv (r"C:\Users\espeo\OneDrive\Escritorio\vientovis\vientovis\vientovis\data\damage\nfip_storm_totals.csv")
```

```python
# formato yy-mm-dd, cantidad solo tiene el año
hurdat2_df["datetime"] = pd.to_datetime(hurdat2_df["datetime"]).dt.strftime(
    "%y-%m-%d")


Impactos["date"] = pd.to_datetime(
    Impactos["date"], format="mixed", errors="coerce"
).dt.strftime("%y-%m-%d")

Dano["date"] = pd.to_datetime(
    Dano["date"], format="mixed", errors="coerce"
).dt.strftime("%y-%m-%d")
```

```python
# Cortar año de columna storm df daño 
Dano["storm"] = (Dano["storm"].str.replace(r"\s*\d{4}", "", regex=True).str.strip().str.upper())
Impactos["storm"] = (Impactos["storm"].str.replace(r"\s*\d{4}", "", regex=True).str.strip().str.upper())
Cantidad["storm"] = Cantidad["storm"].astype(str).str.strip()
Cantidad["datetime"] = Cantidad["storm"].str.extract(r"(\d{4})")
Cantidad["storm"] = (Cantidad["storm"].str.replace(r"\d{4}", "", regex=True).str.strip().str.upper())
Cantidad
```

```python
hurdat2_df.rename(columns={"storm_name": "storm"}, inplace=True)
Impactos.rename(columns={"date": "datetime"}, inplace=True)
Dano.rename(columns={"date": "datetime"}, inplace=True)
```

```python
# Concatenar Impactos a hurdat2_df

impactos_resumen = Impactos.groupby(
    ["storm", "datetime"], as_index=False
).first()
dano_resumen = Dano.groupby(["storm", "datetime"], as_index=False).first()

# storm y datetime
cols_impactos = [
    col
    for col in impactos_resumen.columns
    if col not in ["storm", "datetime"] and col not in hurdat2_df.columns
]
cols_dano = [
    col
    for col in dano_resumen.columns
    if col not in ["storm", "datetime"]
    and col not in hurdat2_df.columns
    and col not in cols_impactos
]

# Unimos con hurdat
hurdat2_df = hurdat2_df.merge(
    impactos_resumen[["storm", "datetime"] + cols_impactos],
    on=["storm", "datetime"],
    how="left",
)

hurdat2_df = hurdat2_df.merge(
    dano_resumen[["storm", "datetime"] + cols_dano],
    on=["storm", "datetime"],
    how="left",
)


hurdat2_df[(hurdat2_df["storm"] == "KATRINA") & (hurdat2_df["date"].str.contains("2005"))]
```

```python
# Sacamos el año de date de hurdat 
hurdat2_df["datetime_year"] = hurdat2_df["date"].astype(str).str[:4]

Cantidad["datetime"] = Cantidad["datetime"].astype(str).str.strip()
Cantidad["storm"] = Cantidad["storm"].astype(str).str.strip().str.upper()
#dejamos todas las columnas mismo nombre
cantidad_renombrado = Cantidad.rename(
    columns={
        "datetime": "year_storm",
        "building_damage_usd": "total_building_damage_usd",
        "contents_damage_usd": "total_contents_damage_usd",
        "api_count": "total_api_count",
        "records_pulled": "total_records_pulled",
        "complete": "total_complete",
    }
)

hurdat2_df = hurdat2_df.merge(
    cantidad_renombrado,
    left_on=["storm", "datetime_year"],
    right_on=["storm", "year_storm"],
    how="inner",
).drop(columns=["datetime_year", "year_storm"])

hurdat2_df

hurdat2_df
```
